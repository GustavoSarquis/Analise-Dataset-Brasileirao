```python
vitorias_casa = (
    df_analise.groupby("mandante")
    .agg(
        jogos_casa=("partida_id", "count"),
        vitorias_casa=("gols_mandante", lambda x: (x > df_analise.loc[x.index, "gols_visitante"]).sum())
    )
)

vitorias_fora = (
    df_analise.groupby("visitante")
    .agg(
        jogos_fora=("partida_id", "count"),
        vitorias_fora=("gols_visitante", lambda x: (x > df_analise.loc[x.index, "gols_mandante"]).sum())
    )
)

comparacao_casa_fora = vitorias_casa.join(vitorias_fora, how="inner")

comparacao_casa_fora["%_vit_casa"] = (
    comparacao_casa_fora["vitorias_casa"] /
    comparacao_casa_fora["jogos_casa"] * 100
)

comparacao_casa_fora["%_vit__fora"] = (
    comparacao_casa_fora["vitorias_fora"] /
    comparacao_casa_fora["jogos_fora"] * 100
)

comparacao_casa_fora["diferenca"] = (
    comparacao_casa_fora["%_vit_casa"] -
    comparacao_casa_fora["%_vit__fora"]
)

comparacao_casa_fora[["%_vit_casa", "%_vit__fora", "diferenca"]] = (
    comparacao_casa_fora[["%_vit_casa", "%_vit__fora", "diferenca"]].round(2)
)

comparacao_casa_fora.sort_values("diferenca", ascending=False)

# Sugestão de gráfico: barras agrupadas comparando a porcentagem de vitórias em casa e fora por time; biblioteca: Plotly.
```

```python
def converter_minuto(valor):
    valor = str(valor)

    if "+" in valor:
        minuto, acrescimo = valor.split("+")
        return int(minuto) + int(acrescimo)

    return int(valor)


gols_analise = gols.copy()

gols_analise["minuto_num"] = gols_analise["minuto"].apply(
    converter_minuto
)
```

```python
def classificar_faixa_minuto(valor):
    valor = str(valor)
    minuto_base = int(valor.split("+")[0])

    if minuto_base <= 15:
        return "0–15"
    elif minuto_base <= 30:
        return "16–30"
    elif minuto_base <= 45:
        return "31–45+"
    elif minuto_base <= 60:
        return "46–60"
    elif minuto_base <= 75:
        return "61–75"
    else:
        return "76–90+"


gols_analise["faixa_minuto"] = gols_analise["minuto"].apply(
    classificar_faixa_minuto
)
```

```python
def contar_outliers(serie):
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)

    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    outliers = serie[
        (serie < limite_inferior) |
        (serie > limite_superior)
    ]

    return {
        "Q1": q1,
        "Q3": q3,
        "Limite inferior": limite_inferior,
        "Limite superior": limite_superior,
        "Quantidade de outliers": len(outliers)
    }
```
